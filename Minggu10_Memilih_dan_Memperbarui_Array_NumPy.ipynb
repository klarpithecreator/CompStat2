## 1. Mengurutkan Array (Sorting)

### 1.1 Pengantar Pengurutan Array

**Sorting** adalah proses mengurutkan elemen array, misalnya dari nilai terkecil ke terbesar. Fungsi `np.sort()` menghasilkan **array baru** yang sudah terurut, sehingga array aslinya **tidak berubah**.

### 1.2 Mengurutkan Array Satu Dimensi (1D)

Untuk array 1D, `np.sort()` mengurutkan semua elemen secara menaik (*ascending*). Gunakan `[::-1]` pada hasilnya jika ingin urutan menurun (*descending*).

```python
scores = np.array([91, 67, 84, 78, 95])

print("Asli:", scores)
print("Menaik:", np.sort(scores))
print("Menurun:", np.sort(scores)[::-1])
print("Array asli tetap:", scores)
```

`[::-1]` → membalik urutan array, sehingga hasil yang semula menaik menjadi menurun.

### 1.3 Mengurutkan Array Dua Dimensi (2D)

Pada array 2D, `np.sort()` secara bawaan mengurutkan nilai **di dalam setiap baris** (`axis=-1`, sama dengan `axis=1` untuk array 2D). Fungsi ini **tidak** mengurutkan seluruh baris berdasarkan satu kolom tertentu.

```python
measurement_matrix = np.array([[7, 2, 9],
                               [5, 8, 1],
                               [6, 4, 3]])

print("Matriks asli:")
print(measurement_matrix)
print("Terurut di dalam setiap baris:")
print(np.sort(measurement_matrix))
```

### 1.4 Mengurutkan Berdasarkan Sumbu (Axis)

Untuk array 2D:
- **`axis=0`**: mengurutkan nilai dari atas ke bawah pada **setiap kolom**.
- **`axis=1`**: mengurutkan nilai dari kiri ke kanan pada **setiap baris**.
- **`axis=None`**: mengurutkan seluruh elemen dan menghasilkan array 1D.

```python
print("Terurut per kolom (axis=0):")
print(np.sort(measurement_matrix, axis=0))

print("Terurut per baris (axis=1):")
print(np.sort(measurement_matrix, axis=1))

print("Seluruh elemen terurut:", np.sort(measurement_matrix, axis=None))
```

### 1.5 Mengurutkan Indeks dengan `np.argsort()`

`np.argsort()` menghasilkan **indeks urutan**, bukan nilai yang langsung diurutkan. Indeks tersebut berguna jika nilai dan nama mahasiswa harus tetap berpasangan — urutkan dulu nilainya, lalu pakai indeks yang sama untuk menata nama.

```python
students = np.array(["Ayu", "Budi", "Citra", "Doni"])
final_scores = np.array([79, 92, 85, 71])
order = np.argsort(final_scores)

print("Nilai terurut:", final_scores[order])
print("Mahasiswa yang bersesuaian:", students[order])
```

```python
order
```

Arti `order`: elemen pertama (`3`) berarti nilai terkecil ada di indeks 3 (Doni), dan seterusnya sampai nilai terbesar.

## 2. Menyaring Array (Filtering)

### 2.1 Memahami Boolean Mask

**Filtering** digunakan untuk memilih data yang memenuhi syarat tertentu. **Boolean mask** adalah array berisi `True` dan `False` yang dihasilkan dari operasi perbandingan.

```python
exam_scores = np.array([58, 83, 90, 76, 69, 95])
mask = exam_scores >= 75

print("Nilai:", exam_scores)
print("Mask >= 75:", mask)
```

### 2.2 Boolean Indexing

Gunakan `array[mask]` untuk mengambil elemen pada posisi yang bernilai `True`. Hasilnya hanya berisi elemen yang memenuhi syarat.

```python
print("Nilai >= 75:", exam_scores[exam_scores >= 75])
print("Nilai < 75:", exam_scores[exam_scores < 75])
```

### 2.3 Boolean Indexing pada Array Dua Dimensi

Untuk matriks, kondisi dapat digunakan untuk memilih **baris** berdasarkan salah satu kolom. Perhatikan bahwa hasil pemilihan beberapa baris tetap berupa array 2D.

```python
student_marks = np.array([[72, 80, 77],
                          [88, 91, 85],
                          [65, 70, 74],
                          [83, 79, 90]])

selected_rows = student_marks[student_marks[:, 0] >= 80]
print("Mahasiswa dengan nilai pertama >= 80:")
print(selected_rows)
print("Bentuk hasil:", selected_rows.shape)
```

```python
student_marks[:, 0]
```

Urutan kerjanya: `student_marks[:, 0]` adalah slicing semua baris (`:`) pada kolom pertama (`0`). Kondisi `>= 80` pada kolom itu menghasilkan Boolean mask, lalu mask tersebut dipakai untuk menyaring **baris-baris** yang sesuai dengan semua kolomnya.

```python
student_marks[student_marks[:, 0] >= 80]
```

### 2.4 Fancy Indexing

**Fancy indexing** memilih elemen menggunakan daftar indeks tertentu. Berbeda dengan Boolean indexing, di sini kita menentukan **posisi** elemen yang ingin diambil, dan urutan indeks menentukan urutan hasil.

```python
values = np.array([5, 15, 25, 35, 45])
print("Posisi terpilih:", values[[1, 3, 4]])

print("Baris terpilih:")
print(student_marks[[1, 2], :])
```

### 2.5 Menyaring dengan `np.where()`

`np.where(kondisi)` menghasilkan **indeks posisi** yang memenuhi kondisi. Sementara itu, `np.where(kondisi, nilai_benar, nilai_salah)` memilih nilai berdasarkan kondisi, bukan hanya menyaring elemen.

```python
rainfall = np.array([3, 15, 0, 22, 9, 30])
print("Indeks dengan curah hujan >= 10:", np.where(rainfall >= 10)[0])
print("Nilai dengan curah hujan >= 10:", rainfall[np.where(rainfall >= 10)])

print("Hujan / Tidak hujan:", np.where(rainfall > 0, "Hujan", "Tidak hujan"))
```

```python
np.where(rainfall >= 10)
```

Perhatikan hasilnya berupa **tuple** berisi satu array (satu array per sumbu). Itulah alasan kita memakai `[0]` untuk mengambil array indeksnya.

### 2.6 Seleksi Bersyarat dengan Banyak Kondisi

Gunakan **`&`** untuk *AND*, **`|`** untuk *OR*, dan **`~`** untuk *NOT*. Setiap kondisi harus diberi tanda kurung. Jangan menggunakan `and` atau `or` untuk membandingkan seluruh array NumPy (akan muncul `ValueError`).

```python
temperatures = np.array([24, 27, 30, 33, 29, 35])

print("27 sampai 31:", temperatures[(temperatures >= 27) & (temperatures <= 31)])
print("Di bawah 27 atau di atas 33:", temperatures[(temperatures < 27) | (temperatures > 33)])
print("Tidak di atas 30:", temperatures[~(temperatures > 30)])
```

## 3. Memperbarui Array (Updating)

### 3.1 Memperbarui Elemen Individual

Nilai array bisa diubah menggunakan indeks, misalnya `array[1] = 10`. Gunakan `.copy()` jika ingin mengubah array baru tanpa mengubah data aslinya.

```python
original_data = np.array([5, 15, 25, 35])
updated_data = original_data.copy()
updated_data[2] = 40

print("Asli:", original_data)
print("Diperbarui:", updated_data)
```

Tanpa `.copy()`, nama variabel baru hanya menjadi "nama lain" untuk array yang sama, sehingga mengubah salah satunya ikut mengubah yang lain. Buktikan sendiri:

```python
alias_data = original_data      # bukan salinan, hanya nama lain
alias_data[0] = 99

print("alias_data   :", alias_data)
print("original_data:", original_data)
```

Nilai `original_data` ikut berubah menjadi `99`. Itulah sebabnya `.copy()` penting untuk menjaga data asli.

### 3.2 Memperbarui Baris dan Kolom

Pada array 2D, gunakan `array[baris, kolom]` untuk satu nilai, `array[baris, :]` untuk satu baris, dan `array[:, kolom]` untuk satu kolom. Jumlah nilai pengganti harus sesuai dengan posisi yang dipilih, kecuali jika memakai satu nilai skalar (yang akan "disebarkan" ke semua posisi terpilih).

```python
marks = np.array([[60, 65, 70],
                  [72, 76, 78],
                  [80, 81, 83]])

marks[0, 0] = 68
marks[1, :] = [74, 77, 80]
marks[:, 2] = 90

print("Nilai setelah diperbarui:")
print(marks)
```

### 3.3 Penggantian Bersyarat

Nilai yang memenuhi kondisi tertentu dapat diganti sekaligus dengan **Boolean indexing** atau `np.where()`. Untuk menghindari perubahan data asli, salin array terlebih dahulu.

```python
measurements = np.array([7, -4, 15, -8, 11])
cleaned = measurements.copy()
cleaned[cleaned < 0] = 0

print("Asli:", measurements)
print("Diperbarui dengan mask:", cleaned)
print("Diperbarui dengan np.where:", np.where(measurements < 0, 0, measurements))
```

`np.where(measurements < 0, 0, measurements)` artinya: jika nilainya `< 0` beri nilai `0`, jika tidak maka kembalikan nilai aslinya. Perbedaannya: `np.where` selalu menghasilkan array **baru**, sedangkan mask dengan `=` mengubah array yang ada.

### 3.4 Menambah Baris dengan `np.concatenate()`

`np.concatenate((a, b), axis=0)` menggabungkan **baris**. Untuk matriks, jumlah **kolom** kedua array harus sama. Fungsi ini menghasilkan array baru.

```python
sales = np.array([[9, 14],
                  [11, 16]])
new_row = np.array([[13, 17]])

sales_with_row = np.concatenate((sales, new_row), axis=0)
print("Setelah menambah baris:")
print(sales_with_row)
print("Bentuk baru:", sales_with_row.shape)
```

### 3.5 Menambah Kolom dengan `np.concatenate()`

Gunakan `axis=1` untuk menggabungkan **kolom**. Jumlah **baris** harus sama. Kolom baru perlu berbentuk 2D, misalnya `(3, 1)`, bukan `(3,)`.

```python
new_column = np.array([[19], [20], [21]])
sales_with_column = np.concatenate((sales_with_row, new_column), axis=1)

print("Setelah menambah kolom:")
print(sales_with_column)
print("Bentuk baru:", sales_with_column.shape)
```

### 3.6 Menghapus Baris dan Kolom dengan `np.delete()`

`np.delete(array, indeks, axis=0)` menghapus **baris**, sedangkan `axis=1` menghapus **kolom**. Hasilnya adalah array baru; array asli tetap tersedia.

```python
print("Hapus baris kedua:")
print(np.delete(sales_with_column, 1, axis=0))
print("Hapus kolom ketiga:")
print(np.delete(sales_with_column, 2, axis=1))
```

```python
sales_with_column
```

Array asli `sales_with_column` tidak berubah setelah penghapusan.

### 3.7 Kompatibilitas Bentuk dan Dimensi

Saat menggabungkan array, dimensi (`ndim`) harus sama. Untuk `axis=0`, jumlah kolom harus sama; untuk `axis=1`, jumlah baris harus sama. `reshape()` membantu mengubah vektor 1D menjadi baris atau kolom 2D.

```python
matrix = np.arange(1, 9).reshape(2, 4)
row_values = np.array([9, 10, 11, 12]).reshape(1, 4)
column_values = np.array([13, 14]).reshape(2, 1)

print("Bentuk matriks:", matrix.shape)
print("Bentuk baris baru:", row_values.shape)
print("Bentuk kolom baru:", column_values.shape)

print("Dengan tambahan baris:")
print(np.concatenate((matrix, row_values), axis=0))

print("Dengan tambahan kolom:")
print(np.concatenate((matrix, column_values), axis=1))
```

**Kesalahan umum:** Menambahkan kolom berbentuk `(2,)` ke matriks `(2, 4)` dengan `axis=1` tidak dapat dilakukan langsung. Ubah dahulu menjadi `(2, 1)`. Jangan pula memakai `np.delete()` tanpa `axis` jika ingin mempertahankan bentuk matriks, karena hasil bawaannya diratakan menjadi 1D:

```python
print(np.delete(matrix, 2))
```

## 4. Agregasi Array

### 4.1 Jumlah, Minimum, Maksimum, dan Rata-rata

**Agregasi** merangkum banyak nilai menjadi satu atau beberapa statistik. Fungsi dasar yang sering dipakai adalah `np.sum()`, `np.min()`, `np.max()`, dan `np.mean()`.

```python
daily_visitors = np.array([110, 125, 140, 132, 118])

print("Total:", np.sum(daily_visitors))
print("Minimum:", np.min(daily_visitors))
print("Maksimum:", np.max(daily_visitors))
print("Rata-rata:", np.mean(daily_visitors))
```

### 4.2 Agregasi Berdasarkan Sumbu

Pada matriks, **`axis=0`** meringkas nilai dari berbagai baris untuk **setiap kolom**; **`axis=1`** meringkas nilai dari berbagai kolom untuk **setiap baris**.

Cara mudah mengingatnya:

- **`axis=0`** → total/rata-rata/... dari semua baris, hasilnya **per kolom** (per variabel).
- **`axis=1`** → total/rata-rata/... dari semua kolom, hasilnya **per baris** (per observasi).

```python
test_scores = np.array([[72, 78, 84],
                        [65, 70, 75],
                        [88, 90, 94]])

print("Total per ujian (axis=0):", np.sum(test_scores, axis=0))
print("Rata-rata per mahasiswa (axis=1):", np.mean(test_scores, axis=1))
print("Maksimum per ujian (axis=0):", np.max(test_scores, axis=0))
```

### 4.3 Jumlah Kumulatif

`np.cumsum()` menghitung jumlah **kumulatif** secara bertahap. Nilai di setiap posisi adalah jumlah seluruh nilai sampai posisi tersebut. `axis` juga dapat digunakan untuk data 2D.

```python
weekly_sales = np.array([8, 14, 11, 20])

print("Penjualan mingguan:", weekly_sales)
print("Penjualan kumulatif:", np.cumsum(weekly_sales))

print("Nilai kumulatif sepanjang kolom:")
print(np.cumsum(test_scores, axis=1))
```

### 4.4 Argumen `keepdims`

Secara bawaan, agregasi menghilangkan sumbu yang diringkas. `keepdims=True` mempertahankan sumbu tersebut dengan ukuran **1**, sehingga hasilnya lebih mudah dipakai untuk operasi lanjutan yang membutuhkan bentuk kompatibel (misalnya broadcasting).

```python
row_means = np.mean(test_scores, axis=1)
row_means_2d = np.mean(test_scores, axis=1, keepdims=True)

print("Tanpa keepdims:", row_means, "bentuk:", row_means.shape)
print("Dengan keepdims:")
print(row_means_2d)
print("Bentuk:", row_means_2d.shape)
```

### 4.5 Menerapkan Agregasi pada Data Statistik

Misalkan setiap baris adalah satu hari dan setiap kolom adalah satu lokasi pengamatan curah hujan. Agregasi dapat menghitung total per lokasi dan rata-rata per hari.

```python
rainfall_matrix = np.array([[9, 14, 6],
                            [12, 8, 10],
                            [18, 13, 7],
                            [5, 9, 3]])

print("Total per lokasi:", np.sum(rainfall_matrix, axis=0))
print("Rata-rata per hari:", np.mean(rainfall_matrix, axis=1))
print("Maksimum keseluruhan:", np.max(rainfall_matrix))
```

## 5. Operasi Vektorisasi

### 5.1 Vektorisasi vs. Perulangan Python

**Vektorisasi** memungkinkan perhitungan langsung pada seluruh array tanpa menulis perulangan untuk setiap elemen. Kodenya umumnya lebih singkat dan sering jauh lebih efisien untuk data numerik.

```python
values = np.array([3, 5, 7, 9])

# Perulangan Python
loop_result = []
for value in values:
    loop_result.append(value * 3)

# Vektorisasi NumPy
vector_result = values * 3

print("Perulangan:", loop_result)
print("Vektorisasi:", vector_result)
```

Catatan: pada NumPy versi 2, elemen list hasil perulangan tampil sebagai `np.int64(...)`. Itu hanya cara tampilan; nilainya tetap angka biasa.

### 5.2 Operasi Aritmetika Skalar

Satu bilangan (**skalar**) dapat dijumlahkan, dikurangkan, dikalikan, atau dibagikan ke setiap elemen array. Operasi ini bekerja untuk array 1D maupun 2D.

```python
data = np.array([5, 15, 25])

print("Tambah 4:", data + 4)
print("Kurang 3:", data - 3)
print("Kali 2:", data * 2)
print("Bagi 5:", data / 5)
print("Kuadrat:", data ** 2)
```

### 5.3 Matematika Array Element-wise

**Element-wise** berarti operasi dilakukan pada elemen yang **posisinya bersesuaian**. Contohnya, `a + b` menjumlahkan elemen pertama dengan pertama, kedua dengan kedua, dan seterusnya. Kedua array pada contoh berikut memiliki bentuk yang sama.

```python
a = np.array([12, 24, 36])
b = np.array([3, 4, 6])

print("Penjumlahan:", a + b)
print("Pengurangan:", a - b)
print("Perkalian:", a * b)
print("Pembagian:", a / b)
print("Pangkat:", a ** 2)
```

### 5.4 Operasi Perbandingan Element-wise

Operator `>`, `<`, `>=`, `<=`, `==`, dan `!=` membandingkan setiap elemen dan menghasilkan array Boolean. Hasil ini dapat dipakai kembali untuk menyaring atau memperbarui nilai.

```python
actual = np.array([82, 70, 91, 65])
target = np.array([80, 75, 85, 65])

print("Aktual >= target:", actual >= target)
print("Aktual == target:", actual == target)
print("Mencapai target:", actual[actual >= target])
```

### 5.5 Kompatibilitas Bentuk dan Broadcasting

Operasi element-wise memerlukan bentuk yang **kompatibel**. **Broadcasting** memungkinkan array dengan bentuk berbeda dipasangkan apabila ukurannya sama atau salah satunya 1 saat dibandingkan dari dimensi paling kanan. Contohnya, vektor tiga nilai dapat ditambahkan pada setiap baris matriks yang memiliki tiga kolom.

```python
matrix = np.array([[15, 25, 35],
                   [45, 55, 65]])
adjustment = np.array([2, 4, 6])

print("Bentuk matriks:", matrix.shape)
print("Bentuk penyesuaian:", adjustment.shape)
print("Hasil element-wise:")
print(matrix + adjustment)
```

**Perbedaan penting:** Operator `*` pada dua array adalah perkalian **element-wise**, bukan perkalian matriks. Untuk perkalian matriks gunakan `@` atau `np.matmul()`. Operasi dengan bentuk yang tidak kompatibel akan menghasilkan `ValueError`. Bandingkan:

```python
p = np.array([[1, 2],
              [3, 4]])
q = np.array([[5, 6],
              [7, 8]])

print("p * q (element-wise):")
print(p * q)
print("p @ q (perkalian matriks):")
print(p @ q)
```

### 5.6 Contoh Terapan: Selisih terhadap Rata-rata

Contoh sederhana berikut menghitung selisih setiap pengamatan dari rata-ratanya. Karena memakai operasi vektorisasi, kita tidak perlu menulis perulangan. Perhatikan bahwa jumlah seluruh selisih terhadap rata-rata selalu (hampir) nol.

```python
observations = np.array([10.0, 14.0, 18.0, 22.0])
mean_value = np.mean(observations)
deviations = observations - mean_value

print("Rata-rata:", mean_value)
print("Selisih:", deviations)
print("Jumlah selisih:", np.sum(deviations))
```

## 7. Contoh Solusi

**Tinjau solusi ini setelah Anda mencoba mengerjakan latihan.** Nama variabel yang berbeda dapat diterima selama nilai dan bentuk yang diminta sudah benar.

### Solusi 1 — Mengurutkan Array

```python
temperatures = np.array([34, 29, 31, 27, 33, 30])

print("Asli       :", temperatures)
print("Menaik     :", np.sort(temperatures))
print("Menurun    :", np.sort(temperatures)[::-1])
print("Asli tetap :", temperatures)

rainfall = np.array([[9, 4, 7],
                     [3, 8, 6],
                     [5, 2, 1]])

print("Terurut per kolom (axis=0):")
print(np.sort(rainfall, axis=0))
print("Terurut per baris (axis=1):")
print(np.sort(rainfall, axis=1))
```

**Perbedaannya:** `axis=0` mengurutkan nilai **di dalam setiap kolom** (dari atas ke bawah), sedangkan `axis=1` mengurutkan nilai **di dalam setiap baris** (dari kiri ke kanan). Keduanya tidak menukar posisi baris atau kolom secara utuh — hanya mengurutkan nilai di sepanjang sumbu yang dipilih.

### Solusi 2 — Menyaring Array

```python
scores = np.array([62, 85, 91, 73, 89, 97, 68])

print("Nilai >= 85:", scores[scores >= 85])
print("Nilai 70 sampai 90:", scores[(scores >= 70) & (scores <= 90)])
print("Indeks 1, 4, 6:", scores[[1, 4, 6]])
print("Indeks nilai > 90:", np.where(scores > 90)[0])
```

### Solusi 3 — Memperbarui Array

```python
inventory = np.array([[14, -2], [9, 11], [16, -5]])

cleaned_inventory = inventory.copy()
cleaned_inventory[cleaned_inventory < 0] = 0
print("Asli:")
print(inventory)
print("Setelah nilai negatif diganti 0:")
print(cleaned_inventory)

inventory_with_row = np.concatenate((cleaned_inventory, np.array([[18, 7]])), axis=0)
print("Setelah menambah baris:")
print(inventory_with_row)
print("Bentuk:", inventory_with_row.shape)

inventory_with_column = np.concatenate((inventory_with_row, np.array([[5], [6], [7], [8]])), axis=1)
print("Setelah menambah kolom:")
print(inventory_with_column)
print("Bentuk:", inventory_with_column.shape)

final_inventory = np.delete(inventory_with_column, 0, axis=0)
final_inventory = np.delete(final_inventory, -1, axis=1)
print("Setelah hapus baris pertama dan kolom terakhir:")
print(final_inventory)
print("Bentuk akhir:", final_inventory.shape)
```

### Solusi 4 — Agregasi Array

```python
visitors = np.array([[22, 28, 35],
                     [19, 24, 30],
                     [26, 29, 34]])

print("Total seluruhnya:", np.sum(visitors))
print("Rata-rata seluruhnya:", np.mean(visitors))
print("Total per kolom (axis=0):", np.sum(visitors, axis=0))
print("Rata-rata per baris (axis=1):", np.mean(visitors, axis=1))
print("Kumulatif per baris:")
print(np.cumsum(visitors, axis=1))

row_means_keep = np.mean(visitors, axis=1, keepdims=True)
print("Rata-rata per baris (keepdims=True):")
print(row_means_keep)
print("Bentuk:", row_means_keep.shape)
```

### Solusi 5 — Operasi Vektorisasi

```python
prices = np.array([12000, 18000, 24000, 30000])
quantities = np.array([4, 2, 3, 1])

revenue = prices * quantities
print("Pendapatan per produk:", revenue)
print("Total pendapatan:", np.sum(revenue))

discounted_prices = prices * 0.85
print("Harga setelah diskon 15%:", discounted_prices)

print("Pendapatan >= 45000:", revenue >= 45000)
```

### Solusi 6 — Praktik Statistik Terpadu

```python
student_results = np.array([[88, 78, 97],
                            [60, 68, 74],
                            [93, 96, 91],
                            [72, 80, 77]])

student_means = np.mean(student_results, axis=1)
print("Rata-rata per mahasiswa:", student_means)

print("Baris dengan rata-rata >= 80:")
print(student_results[student_means >= 80])

ranking = np.argsort(student_means)
print("Urutan indeks dari rata-rata terendah ke tertinggi:", ranking)
print("Posisi (indeks) mahasiswa dengan rata-rata tertinggi:", ranking[-1])

boosted = student_results + 5
updated_results = np.where(boosted > 100, 100, boosted)
print("Nilai setelah +5 (maksimum 100):")
print(updated_results)

print("Rata-rata per ujian setelah pembaruan:", np.mean(updated_results, axis=0))
```

## 8. Ringkasan Pertemuan dan Penilaian Mandiri

### Konsep Penting

- **`np.sort()`** mengurutkan nilai dan **`np.argsort()`** menghasilkan indeks pengurutan.
- **`axis=0`** bekerja sepanjang baris untuk setiap kolom; **`axis=1`** bekerja sepanjang kolom untuk setiap baris.
- **Boolean mask**, **Fancy indexing**, dan **`np.where()`** membantu memilih data sesuai kondisi atau posisi.
- **Penugasan indeks** (*index assignment*) dan **penggantian bersyarat** digunakan untuk memperbarui nilai array; gunakan `.copy()` agar data asli aman.
- **`np.concatenate()`** menambahkan baris/kolom dan **`np.delete()`** menghapusnya dengan `axis` yang sesuai.
- **`.ndim`** dan **`.shape`** penting untuk memeriksa kompatibilitas bentuk saat menggabungkan array.
- **`np.sum()`**, **`np.min()`**, **`np.max()`**, dan **`np.mean()`** menghitung ringkasan data.
- **`np.cumsum()`** menghitung jumlah kumulatif; **`keepdims=True`** mempertahankan dimensi yang diringkas.
- **Vektorisasi** memungkinkan operasi skalar dan element-wise tanpa perulangan manual; perbandingan menghasilkan array Boolean.

### Penilaian Mandiri Singkat

Sebelum pertemuan berikutnya, jawab pertanyaan berikut tanpa menjalankan kode:

1. Apa perbedaan hasil `np.sort(x, axis=0)` dan `np.sort(x, axis=1)` pada array 2D?
2. Apa hasil `np.array([4, 9, 14, 19])[np.array([4, 9, 14, 19]) >= 14]`?
3. Apa perbedaan `np.where(a > 0)` dan `np.where(a > 0, a, 0)`?
4. Berapa bentuk hasil penggabungan array `(3, 4)` dan `(1, 4)` jika `axis=0`?
5. Berapa bentuk hasil `np.mean(np.zeros((5, 2)), axis=1, keepdims=True)`?
6. Apa hasil `np.array([3, 5, 7]) * np.array([2, 3, 1])`?
7. Apa hasil perbandingan `np.array([3, 9, 15]) >= 9`?

**Jawaban:** (1) `axis=0` mengurutkan setiap kolom, `axis=1` mengurutkan setiap baris; (2) `[14, 19]`; (3) yang pertama menghasilkan indeks dan yang kedua menghasilkan nilai berdasarkan kondisi; (4) `(4, 4)`; (5) `(5, 1)`; (6) `[6, 15, 7]`; (7) `[False, True, True]`.

Cek jawabannya dengan kode:

```python
print(np.array([4, 9, 14, 19])[np.array([4, 9, 14, 19]) >= 14])
print(np.concatenate((np.zeros((3, 4)), np.zeros((1, 4))), axis=0).shape)
print(np.mean(np.zeros((5, 2)), axis=1, keepdims=True).shape)
print(np.array([3, 5, 7]) * np.array([2, 3, 1]))
print(np.array([3, 9, 15]) >= 9)
```
